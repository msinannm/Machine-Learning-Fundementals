```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\MERIN\2. Finance_Analytics_Dataset.csv")

df
```

```python
# 1

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Category")
plt.title("Number of Transactions in Each Category")
plt.xlabel("Category")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.show()
```

```python
# 2

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="PaymentMode")
plt.title("Number of Transactions by Payment Mode")
plt.xlabel("Payment Mode")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.show()
```

```python
# 3. 

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="TransactionType")
plt.title("Number of Transactions by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
# 4.

df.groupby("TransactionType")["Amount"].sum().plot(kind="bar")
plt.title("Debit vs Credit Amount")
plt.ylabel("Amount")
plt.show()
```

```python
# 5. 

df.groupby("CustomerSegment")["Amount"].sum().plot(kind="bar")
plt.title("Spending by Customer Segment")
plt.ylabel("Total Amount")
plt.show()
```

```python
# 6. 

df["PaymentMode"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Payment Mode Distribution")
plt.ylabel("")
plt.show()
```

```python
# 7. 

sns.boxplot(x="Category", y="Amount", data=df)
plt.title("Transaction Amount by Category")
plt.xticks(rotation=45)
plt.show()
```
